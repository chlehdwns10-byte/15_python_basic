```python
#딕셔너리 키의 자료형 
## 1. 문자형 가능 
## 2. 정수형 가능
## 3. 실수형 가능
## 4. 불(True, False) 가능

##불가능한 키의 자료형
## 1. 딕셔너리 불가능
## 2. 리스트 불가능 

x = {
    100 : "hundred", 
    False : 0,
    3.5 : [3.5, 3.5]
}

print(type(x))
print(x)
```

```python
# 빈 딕셔너리 만들기
x = {}
y = dict()
print(x)
print(type(x))
print(bool(x))
print(y)
print(type(y))
print(bool(y))
```

```python
# dict 딕셔너리 만들기
lux1 = dict(health = 400, mane =334, melee= 50, armor=10.72)
print(lux1)
print(type(lux1))
```

```python
lux2 = dict(zip(["health", "mane", "melee", "armor"], [490, 334, 550,10.72]))
print(lux2)
print(type(lux2))
```

```python
lux3 = dict([("health", 490),("mane", 334), ("melee", 550), ("armor", 10.72)])
print(lux3)
print(type(lux3))
```

```python
## for range 사용하기
#for 변수 in range(횟수):
#    반복할 코드

for i in range(5,12):
    print("hello world", i)
```

```python
for imsi in range(0,10,2):
    print("hello World", imsi)
```

```python
for a in range(10,0,-1):
    print("hello World", a)
```

```python
print(list(range(10)))
print(list(reversed(range(10))))
```

```python
for c in reversed(range(10)):
    print("hello World", c)
```

```python
#시퀀스 ㄱㄱ체로 반복하기
a = [10, 20, 30, 40, 50]
for i in a:
    print(i)
```

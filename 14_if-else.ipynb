```python
scores = list(map(int,input("점수를 입력하세요: ").split(",")))
average = sum(scores) / len(scores)
if max(scores) > 100 or min(scores) < 0:

    print("잘못된 점수입니다.")
elif average >= 80:
    print("합격입니다.")
else:
    print("불합격입니다.")


             
        
```

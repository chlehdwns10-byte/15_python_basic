```python
#문제 1.
# #input() 함수를 이용하여 국어, 영어, 수학, 과학, 컴퓨터 점수를 한 줄로 입력받으세요.
# 입력된 점수는 콤마 ,로 구분되어 있습니다.
# 입력받은 점수들의 평균 점수를 출력하는 프로그램을 작성하세요
# 
# 입력값 : 
# 80,90,70,85,95
# 
# 출력값 : 
# 평균 점수: 84.0
# 
# 요구사항 
# 1. input() 함수를 사용하세요.
# 2. 점수는 콤마 ,로 구분하여 입력받으세요.
# 3. split()을 사용하여 점수를 분리하세요.

a = input("국어, 영어, 수학, 과학, 컴퓨터")
a_1 = a.split(",")
a_list = list(map(float, a_1))
avg = sum(a_list) / len(a_list)
print(avg)
```

```python
#문제 2.
#input() 함수를 이용하여 년, 월, 일, 시, 분, 초가 입력을 하고 
#제시된 결과값을 출력하는 프로그램을 작상하세요 
#
#입력값 : 
#2026 5 13 16 40 1
#
#출력값 : 
#2026/05/13T16:40:01
#
#요구사항
#1. print() 함수의 매개변수를 활용하세요 

year, month, days, hour, minute, second = input().split()

print(year, month, days, sep="/", end="T")
print(hour, minute, second, sep=":")
```

```python
#문제 3. 
#input() 함수를 이용하여 국어, 영어, 수학, 과학, 컴퓨터 점수를 입력받아 
#한 과목이라도 60점 미만이면 False 가 나오도록 프로그램을 작성하세요 
#
#입력값 : 
#70 80 55 60 90
#
#출력값 : 
#False 
#
#요구사항
#1. 비교연산자와 논리연산자를 사용하세요 

국어, 영어, 수학, 과학 , 컴퓨터 = map(int, input("점수를 입력하세요: "). split())
if 국어 < 60 or 영어 < 60 or 수학 < 60 or 과학 < 60 or 컴퓨터 < 60:
    print(False)
else:
    print(True)
```

```python
#문제 5. 
#range()로 숫자 리스트 만들기
#다음 조건을 만족하는 리스트를 각각 만드세요.
#
#1. 1부터 10까지의 숫자 리스트
#2. 0부터 20까지의 짝수 리스트
#3. 1부터 19까지의 홀수 리스트
#4. 10부터 1까지 거꾸로 된 리스트
#5. 5부터 50까지 5씩 증가하는 리스트


print(list(range(0,11)))
print(list(range(0,21,2)))
print(list(range(1,20,1)))
print(list(range(10,0,-1)))
print(list(range(5,51,5)))
```

```python
#문제 6. 
#input 함수로 입력을 받아서 
#튜플로 출력하는 프로그램을 만드세요 
#
#입력 :
#시작값: 3
#끝값: 15
#증가값: 3
#
#
#출력값 : 
#(3, 6, 9, 12)
#
#요구사항
#1. input(), range(), tuple(), int() 사용하세요


a = int(input("시작값"))
b = int(input("끝값"))
c = int(input("증가값"))
        
tuple(range(a,b,c))
```

```python
#문제 7. 
#이메일 주소를 분리하여 출력하는 프로그램을 만드세요 
#
#입력값 : 
#himzei@gmail.com
#
#출력값 : 
#아이디는 himzei 입니다. 
#도메인은 gmail.com 입니다. 
#
#요구사항 
#1. input(), split() 
#2. 이스케이프 코드를 사용하세요

email = input("이메일 주소 입력해주세요:".split("@"))
print(f"아이디: {email[0]}입니다.\n도메인: {email[1]}")
```

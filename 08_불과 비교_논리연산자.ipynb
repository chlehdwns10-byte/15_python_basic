```python
# 논리연산자와 비교연산자
print((10 == 10) and (10 != 5))
print((10 > 5) or (10 < 3))

print(not 10 > 5)
print(not 1 is 1.0) #  is 이건 다르다는 말임 그래서 1이랑 1.0이랑 다르다는 말
```

```python
(not 10 > 5)  
(not 10) == 1  # 이거 not 10이 0으로 받아들여서 1이랑 같지 않음 (여기서 1은 참 거짓의 1,   그래서 참은 1 거짓은 0)
```

```python
# 변수이름 
## 영문 문자와 숫자를 사용가능
## 대소무낮 구분한다 
## _ (밑줄문자)로 사용할 수 있음
## 특수문자(+. -, /, @, &, %)는 사용할 수 없음
## 파이썬 키워드(if, for, while, def, class 등)는 사용할 수 없음
```

```python
#  변수의 자료형 알아내기 
y = 'hello world'
x = 10

print (type(x))
print (type(y))
```

```python
# 변수 여러 개를 한 번에 만들기

x, y, z = 10, 20, 30
print(x)
print(y)
print(z)
```

```python

x = y = z = 10
print(x)
print(y)
print(z)
```

```python
x, y = 10 ,20  # 초기 설정
x, y = y, x   # 하지만 x,y 값을 서로 바꾼다는 말임
print(x)
print(y)
```

```python
# 변수 계산하기

a = '10'
b = '20' # 이건 문자열임

c =a + b #결과는 1020임
print(c)
```

```python
#  산술 연산 후 할당 연산자 사용하기

a = 10
a + 20
```

```python
a = 10
a = a + 20 # 오른쪽 코드를 먼저 읽고 그 다음 왼쪽 코드로 바뀐다 이런느낌으로
a
```

```python
a= 10
a += 20 # a = a + 20 과 같은 말임 
a
```

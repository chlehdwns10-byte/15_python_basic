```python
x, y = 10 ,20  # 초기 설정
x, y = y, x   # 하지만 x,y 값을 서로 바꾼다는 말임
print(x)
print(y)
```

```python
# 'hello python'.split()
a, b = ['hello', 'python', 'world'] # 두개를 받을건데 3개가 있어서 뭔지 몰라서 오류발생
print(a)
print(b)
```

```python
#  두 숫자의 합 구하기
a, b = input("숫자 2개를 입력하세요:").split()
print(a)
print(b)
print(a + b)
```

```python
#  두 숫자의 합 구하기
a, b = input("숫자 2개를 입력하세요:").split()
a = int(a)
b = int(b)
print(a + b)
```

```python
# 입력받은 값을 콤마를 기준으로 분리하기
a,b = map(int, input("숫자 두 개를 입력하세요: ").split())

#a,b = map(int, ["10" , "20"])

#a,b = 10, 20
print(a + b)
```

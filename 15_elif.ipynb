```python
#elif 사용하기
x = 20
if x == 10:
    print("10입니다.")
elif x== 20:
    print("20입니다.")
```

```python
#elif 사용하기
x = 30
if x == 10:
    print("10입니다.")
elif x== 20:
    print("20입니다.")
else:
    print("30입니다.")
```

```python
# 음료수 자판기
button = int(input())
if button ==1:
    print("콜라")
elif button ==2:
    print("사이다")
elif button ==3:
    print("환타")
else:
    print("제공하지 않는 메뉴")
```

```python
age = int(input("나이를 입력해주세요: "))
balance = 9000
if age < 7:
    print("다시 입력해주시기 바랍니다.")
elif 7 <= age <= 12:
    print(f"{balance-650}원")
elif 13<= age <= 18:
    print(f"{balance -1050}원")
elif age >= 19:
    print(f"{balance -1250}원")

```
